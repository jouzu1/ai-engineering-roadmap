```python
# Import dasar
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 120)
print(np.__version__, pd.__version__)
```

## NumPy singkat

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print("a+b:", a + b)
print("a*2:", a * 2)
print("mean(a):", a.mean())

M = np.array([[1,2],[3,4]])
print("M shape:", M.shape)
print("M @ M:\n", M @ M)
```

## Bangun DataFrame contoh (tabular sintetis)

```python
rng = np.random.default_rng(42)
n = 12
df = pd.DataFrame({
    "city": rng.choice(["A","B","C"], size=n, p=[0.4,0.4,0.2]),
    "area": rng.normal(80, 30, size=n).clip(20, 300),
    "rooms": rng.integers(1, 6, size=n),
    "price": rng.normal(250_000, 60_000, size=n).clip(80_000, None),
})
df.loc[rng.integers(0, n), "area"] = np.nan  # sisipkan 1 missing contoh
df.head()
```

## Ringkasan dan seleksi baris/kolom

```python
display(df.info())
display(df.describe())

# Seleksi kolom
display(df[["city", "area", "rooms"]].head())

# Filter baris
display(df[df["city"] == "A"])  # hanya city A
display(df[(df["area"] > 90) & (df["rooms"] >= 3)])
```

## Agregasi dan groupby

```python
avg_price = df.groupby("city")["price"].mean().reset_index(name="avg_price")
avg_price
```

## Tangani missing values dan duplikat

```python
print("Missing per kolom:\n", df.isna().sum())
df["area"] = df["area"].fillna(df["area"].median())  # imputasi median
df = df.drop_duplicates()
df.isna().sum()
```

## Feature engineering sederhana

```python
df["price_per_m2"] = df["price"] / df["area"]
bins = [0, 70, 100, float("inf")]
labels = ["small", "medium", "large"]
df["area_bin"] = pd.cut(df["area"], bins=bins, labels=labels)
df.head()
```

## Simpan hasil ke datasets/processed (untuk tahap ML selanjutnya)

```python
from pathlib import Path
out_dir = Path("ai-engineering-roadmap/datasets/processed")
out_dir.mkdir(parents=True, exist_ok=True)
out_path = out_dir / "housing_clean.csv"
df.to_csv(out_path, index=False)
out_path
```
